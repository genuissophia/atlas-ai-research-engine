```python

def greet(name):
    return {
        "message": f"Hello, {name}!",
        "status": "success"
    }

result = greet("Sophia")
print(result)
```

```python

from fastapi import FastAPI
from fastapi.testclient import TestClient

app = FastAPI()

@app.get("/greet")
def greet(name: str):
    return {
        "message": f"Hello, {name}!",
        "status": "success"
    }

client = TestClient(app)

response = client.get(
    "/greet",
    params={"name": "Sophia"}
)

print(response.status_code)
print(response.json())
```
